## LinePlot

```python
import numpy as np
```

```python
over=np.arange(1,21)
team_1=np.random.randint(0,37,20)
team_2=np.random.randint(0,37,20)
```

```python
plt.figure(figsize=(7,3))
plt.style.use('dark_background')
plt.plot(over,team1_score,marker='o',color='blue',label='team1')
plt.plot(over,team2_score,marker='<',color='red',label='team2')
plt.xlabel('Overs')
plt.ylabel('Score')
plt.title('Match Summery')
plt.legend()
plt.show()
```
