```python
import matplotlib.pyplot as plt
```

## Bar Plot
1.Compairing categories  
2.Showing frequencies or count

```python
products = ['Hot cofee','Cold cofee','Tea','Salad']
summer=[30,90,80,50]
```

```python
plt.figure(figsize=(7,3)) # width,height
plt.bar(products,summer,color='yellow')
plt.title('Sales of product for summer')
plt.xlabel('Products')
plt.ylabel('Count')
plt.show()
```
